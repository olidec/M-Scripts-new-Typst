## Setup


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

## Part A — pmf and cdf are binompdf and binomcdf

`binom.pmf(k, n, p)` is your calculator's `binompdf`.
`binom.cdf(k, n, p)` is `binomcdf`, and means p(X <= k) — never p(X < k).

The worked sum from the notes: 20 tosses of a fair coin, at least 15 heads.


```python
n, p = 20, 0.5

print("p(X = 15)   ", round(binom.pmf(15, n, p), 5))
print("p(X <= 14)  ", round(binom.cdf(14, n, p), 5))
print("p(X >= 15)  ", round(1 - binom.cdf(14, n, p), 5))
print("the WRONG one, 1 - cdf(15):", round(1 - binom.cdf(15, n, p), 5))
```

**What you should see.** 0.02069 for the right answer and 0.00591 for the
off-by-one. Both are small, both look plausible, and only one is correct.
That is why the notes tell you to write down which k you need before
touching the calculator.


## Part B — the shape

### THE ONE LINE TO CHANGE

Change `n` and `p`. Try (10, 0.5), then (10, 0.2), then (30, 0.2), then
(200, 0.2). Watch the skew disappear as n grows.


```python
n, p = 10, 0.2       # <-- change these two numbers

k = np.arange(0, n + 1)
plt.bar(k, binom.pmf(k, n, p))
plt.axvline(n * p, color="crimson", linestyle="--", label="E(X) = np")
plt.xlabel("number of successes k")
plt.ylabel("p(X = k)")
plt.legend()
plt.show()

print("E(X) =", n * p, "  sigma =", round((n * p * (1 - p)) ** 0.5, 3))
```

**What you should see.** For p = 0.5 the picture is symmetric. For p = 0.2
it leans left with a tail to the right — and that lean fades as n rises,
until at n = 200 it is hard to see any asymmetry at all.

That disappearing skew is the reason the normal distribution exists. Keep
the n = 200 picture in mind for two chapters' time.


## Going further


### G1 — simulate instead of computing

Everything above used the formula. Here the trials are actually run.


```python
TRIALS = 20000
n, p = 10, 0.2

rng = np.random.default_rng(seed=3)
counts = rng.binomial(n, p, size=TRIALS)

k = np.arange(0, n + 1)
plt.hist(counts, bins=np.arange(-0.5, n + 1.5), density=True,
         alpha=0.6, label="simulated")
plt.plot(k, binom.pmf(k, n, p), "o-", color="crimson", label="exact")
plt.legend()
plt.show()

print("simulated mean", round(counts.mean(), 3), " exact", n * p)
```

Drop `TRIALS` to 200 and run it again a few times. The bars stop matching
the dots. Nothing is wrong with the formula — there is simply not enough
data, which is the law of large numbers making the same point for the third
time in this unit.


### G2 — how far off can a poll be?

The poll exercise from the notes: 20 adults, 60% support in the population.
Simulate ten thousand such polls and look at how much the *observed*
percentage moves around.


```python
polls = rng.binomial(20, 0.6, size=10000) / 20

plt.hist(polls, bins=np.arange(-0.025, 1.05, 0.05), edgecolor="white")
plt.axvline(0.6, color="crimson", label="true value 0.6")
plt.xlabel("percentage supporting, in one poll of 20")
plt.legend()
plt.show()

print("polls landing between 0.5 and 0.7:",
      round(((polls >= 0.5) & (polls <= 0.7)).mean(), 3))
```

**What you should see.** A poll of 20 people quite often reports 45% or
75% when the truth is 60%. Now change the sample size from 20 to 1000 and
run it again.

This is the whole argument of the confidence-interval chapter, three
chapters early: a sample tells you about the population, but only to within
a margin, and the margin depends on the sample size.
