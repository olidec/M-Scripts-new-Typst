### Mean against median

The four lines from the lecture notes:


```python
times = df["reaction_ms"]
print("mean  ", round(times.mean(), 1))
print("median", times.median())
print("sd    ", round(times.std(), 1))
```

**What you should see.** The mean is larger than the median. The histogram
below says why: a long tail to the right, and no tail at all to the left.


```python
plt.hist(times, bins=16, edgecolor="white")
plt.axvline(times.mean(), color="crimson", label="mean")
plt.axvline(times.median(), color="black", label="median")
plt.xlabel("reaction time in ms")
plt.ylabel("count")
plt.legend()
plt.show()
```

## Part B — the expected value is a long-run average

A fair die has expected value 3.5. That is not a value the die can show, so
what does it mean? Roll one many times and watch the running average.

### THE ONE LINE TO CHANGE

`N` is the number of rolls. Try 20, then 200, then 20000.


```python
N = 200          # <-- change this one number

rng = np.random.default_rng(seed=1)
rolls = rng.integers(1, 7, size=N)
running = np.cumsum(rolls) / np.arange(1, N + 1)

print("average of", N, "rolls:", round(running[-1], 4))
```

```python
plt.plot(running, linewidth=1)
plt.axhline(3.5, color="crimson", linestyle="--", label="E(X) = 3.5")
plt.xlabel("number of rolls")
plt.ylabel("running average")
plt.legend()
plt.show()
```

**What you should see.** For small `N` the line wanders a long way from 3.5.
For large `N` it settles onto it and stays. That is the law of large numbers,
and it is the reason the expected value is worth computing.

Notice what does *not* happen: the line never becomes flat, and no single
roll is ever 3.5.

### G1 — is one person's reaction time a single number?

Each student made three attempts. If a person had one true reaction time,
the three attempts would differ only a little. Split the data by student and
look.


```python
per_student = df.groupby("student_id")["reaction_ms"]
summary = per_student.agg(["mean", "std", "min", "max"])
summary.head(10)
```

```python
plt.scatter(summary["mean"], summary["std"])
plt.xlabel("student mean in ms")
plt.ylabel("student standard deviation in ms")
plt.show()
```

Two things worth arguing about with your neighbour:

- How does the spread *within* one student compare with the spread *between*
  students? Which is bigger?
- Students with a slower mean also tend to be more variable. Is that a fact
  about people, or a consequence of reaction times having a floor and no
  ceiling?


### G2 — checking an expected value by simulation

The urn game from the lecture notes: ten cards, two of them O, four draws
with replacement, CHF 5 per O, CHF 5 to play. The notes say the expected
net result is CHF -1. Simulate a hundred thousand games and see.


```python
GAMES = 100000

rng = np.random.default_rng(seed=7)
draws = rng.random((GAMES, 4)) < 0.2
payout = 5 * draws.sum(axis=1)
net = payout - 5

print("simulated mean net:", round(net.mean(), 4))
print("theory:            -1")
```

**What you should see.** Something within a couple of centimes of -1.

Then change `GAMES` to 100 and run it again a few times. The simulation is
only as good as its number of repetitions — which is the same lesson as
Part B, arriving from the other side.
